```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver
from langchain.agents.middleware import SummarizationMiddleware

agent = create_agent(
    model="claude-sonnet-4-5-20250929",
    checkpointer=InMemorySaver(),
    middleware=[
        SummarizationMiddleware(
            model="claude-haiku-4-5", # switched to a cheaper model for summarization
            trigger=("tokens", 100),
            keep=("messages", 1)
        )
    ],
)
```

```python
# to test summarization we pass AI messages to the agent as conversation history
from langchain.messages import HumanMessage, AIMessage
from pprint import pprint

response = agent.invoke(
    {"messages": [
        HumanMessage(content="What is the name of my cat?"),
        AIMessage(content="The name of your cat is Casper."),
        HumanMessage(content="What breed was the cat?"),
        AIMessage(content="Casper was Persian and Mainecoon mixed breed."),
        HumanMessage(content="What was the gender of the cat?"),
        AIMessage(content="Casper was a male cat."),
        HumanMessage(content="What color was he?"),
        AIMessage(content="He had a white furcoat with golden brown ears."),
        HumanMessage(content="Describe him to me?")
    ]},
    {"configurable": {"thread_id": "1"}}
)

pprint(response)
```
