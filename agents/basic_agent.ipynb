```python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

agent = create_agent(
    model="claude-sonnet-4-5-20250929",
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)

# Run the agent
agent.invoke(
    {"messages": [{"role": "user", "content": "what is the weather in sf"}]}
)
```

```python
# new creation cuz unless specific the above does not work. 
from langchain.agents import create_agent
import json 
from difflib import get_close_matches

filename = '../resources/coffee.json'

def coffee_recipe(coffee_recipe: str) -> dict:
    """Gets recipe from JSON using natural language input with accurate brew methods."""
    with open(filename, 'r') as file: 
        data = json.load(file)

    recipes = data["recipes"]

    recipe_ids = [r["id"] for r in recipes]

    coffee = coffee_recipe.lower()

    for rid in recipe_ids:
        if rid.lower() in coffee:
            return next(r for r in recipes if r["id"] == rid)
    
    matches = get_close_matches(coffee_recipe, recipe_ids, n=1, cutoff=0.4)

    if matches:
        matched_id = matches[0]
        return next(r for r in recipes if r["id"] == matched_id)
    
    return {"error": "recipe not found"}

agent = create_agent(
    model="claude-sonnet-4-5-20250929",
    tools=[coffee_recipe],
    system_prompt="You are a professional barista."
)

agent.invoke(
    {"messages": [{"role": "user", "content": "Tell me how to brew a kalita"}]}
)
```
